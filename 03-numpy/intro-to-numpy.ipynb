```python
# Broadcasting in NumPy
a = np.array([1, 2, 3])
b = 2
result = a + b
print(result)


#Explanation       
# a is shape (3,)
# b is a single number → shape ()
# NumPy “stretches” b to [2, 2, 2] internally (without copying)
# Then adds element-wise: [1+2, 2+2, 3+2] 
```

```python
# Variance = measure of the average degree to which each number is different to mean
# Higher variance = wider range of numbers
# Lower variance = lower range of numbers
np.var(a2)
```

```python
# Standard deviation = squareroot of variance (A measure of how spread out a group  of numbers is from the mean)
np.std(a2) 
# Both are same 
np.sqrt(np.var(a2))
```

```python
# demo of std and var
high_var_array=np.array([1,100,200,300,4000,5000])
low_var_array=np.array([1,2,4,6,8,10])
```

```python
np.var(high_var_array),np.var(low_var_array)
```

```python
np.std(high_var_array),np.std(low_var_array)
```

```python
np.mean(high_var_array),np.mean(low_var_array)
```
